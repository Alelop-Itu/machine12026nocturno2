```python
import numpy as np

#VarianceThreshold Elimina las columnas constantes que no aportan información para distinguir muestras
from sklearn.feature_selection import VarianceThreshold
```

```python
# Carga de datos.
X = [[0, 0, 1], [0, 1, 0], [1, 0, 0],
[0, 1, 1], [0, 1, 0], [0, 1, 1]]

print(X)
```

```python
# Limpieza de datos: selección de atributos.
# fijamos un umbral (th) para decidir qué porcentaje mínimo deben variar los valores de una columna para considerarse
# una columna relevante a la hora de mantenerla en nuestro estudio
#1.Fijar la varianza para conservar columnas
th = 0.2
#2. Se crea el selector
sel = VarianceThreshold(th)
#Fit calcula la varianza de cada columna y transform elimina las que no superan el umbral
print(sel.fit_transform(X))
#Var = p · (1 − p)
```
