```python
import numpy as np

#clase para rellenar huecos con un valor calculado(media, mediana, moda o una constante)
from sklearn.impute import SimpleImputer 
```

```python
# Carga de datos de entrenamiento
#La media se calcula con las columnas (atributos) sin considerar los NaN
X_train = [[1, 4],
[np.nan, 3],
[7, 8]]
```

```python
# Limpieza de datos: imputación valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizamos la estrategia de la media
#Se usa la media en cada columna para rellanar los huecos. Por defecto buscs np.nan

# Aplicar los métodos "fit" y "transform" para imputar los valores pérdidos de X_train
#fit aprende de los datos y calcula la media de cada columna, ignorando los NaN
#transform sustituye cada NaN por la media aprendida de su columna. 
X_train_clean = imputer.fit_transform(X_train)

#Todos los valores salen como float. porque el imputador devuelve un array de NumPy de tipo flotante- Por eso (1. , 4.)
print(X_train_clean)
```

```python
# Imputación de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
[6, np.nan],
[8, 2]]

# Imputar los datos del conjunto de test
#Sobre test, nunca se usa fit, los hucos se rellenan con las medias aprendidas en entrenamiento y no con las de test
X_test_clean=imputer.transform(X_test)
print(X_test_clean)
```
