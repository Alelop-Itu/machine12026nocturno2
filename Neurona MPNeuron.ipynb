1. Implementación de la MPNeuron


```python
import numpy as np

class MPNeuron:
   def __init__(self):
      self.threshold = None
       
   def model(self, x):
      # input: [1, 0, 1, 0] [x1, x2, .., xn]
      z = sum(x)
      return (z >= self.threshold)

   def predict(self, X):
      # input: [[1, 0, 1, 0], [1, 0, 1, 1]]
      Y = []
      for x in X:
        result = self.model(x)
        Y.append(result)
      return np.array(Y)
```
