```python
from fastapi import FastAPI
app = FastAPI() #(crea la api)

@app.get("/") #(ruta raíz que responde las solicitudes GET y devuelve un JSON)
def inicio():
    return {"mensaje": "Hola mundo"}
```

```python
inicio() #Prueba primera ruta
```

```python
@app.get("/saludar") #Creación de otra ruta
def saludar():
    return {"mensaje": "Bienvenida a FastAPI"}
```

```python
saludar()
```

```python
for ruta in app.routes: #Visualizar las rutas creadas
    print(ruta.path)
```
