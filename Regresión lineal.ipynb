```python
#Importamos las librerias 
#Usado para crear el conjunto de datos como tablas (dataframe)
import pandas as pd
#LinearRegression es el modelo de regresión lineal por m+inimos cuadrados
from sklearn.linear_model import LinearRegression
```

```python
#Creamos el conjunto de datos 
df = pd.DataFrame({
"horas":[1,2,3,4,5],
"nota":[2,3,5,4,6]
})

X= df[["horas"]] # el doble corchete se usa para que la respuesta sea [1,3] y no 1,3
y= df[["nota"]]
print (X)
```

```python
#Creamos nuestro modelo - entrenamiento
#1. Se crea el modelo vacío
modelo = LinearRegression()
#2. Se entrena al modelo, busca la resta y = b₀ + b₁·x , lo que minimiza la suma de los errores al cuadrado entre las notas reales a las predichas
modelo.fit(X, y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Realizamos la nueva predicción
prediccion = modelo.predict([[6]])
print("Predcción: ", prediccion)
```

```python
#Graficar con matplolib el modelo y verificar si el nuevo valor corresponde a la predicción en la recta.
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creando los puntos de la recta
X_linea = np.linspace(1,6,100)
X_linea_df = pd.DataFrame({"horas" : X_linea})
y_linea = modelo.predict(X_linea_df)
```

```python
# Nuevo valor: 6 horas (con nombre de columna, sin warning)
X_nuevo = pd.DataFrame({"horas": [6]})
y_nuevo = modelo.predict(X_nuevo)

# Gráfica
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color="blue", label="Datos reales")
plt.plot(X_linea, y_linea, color="red", label="Recta")
plt.scatter(X_nuevo, y_nuevo, color="green", marker="*", s=200,
            label=f"Predicción 6 h = {y_nuevo[0][0]:.1f}")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Regresión lineal: horas vs nota")
plt.legend()
plt.grid(True)
plt.show()

# Verificación numérica
y_recta_6 = modelo.intercept_[0] + modelo.coef_[0][0] * 6
print("Valor de la recta en x=6:", y_recta_6)
print("¿Coincide con la predicción?", np.isclose(y_recta_6, y_nuevo[0][0]))
```
